```python
import os
from typing import List
from contextlib import asynccontextmanager
from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, Field
import pandas as pd
import numpy as np
import mlflow.sklearn
os.getcwd()
```

```python
MODEL_NAME = "Traffic_Volume_Regressor"
model = None

# ==============================================================================
# 1. MODERN FASTAPI LIFESPAN EVENT HANDLER (Replaces @app.on_event("startup"))
# ==============================================================================
@asynccontextmanager
async def lifespan(app: FastAPI):
    global model
    try:
        model_uri = f"models:/{MODEL_NAME}/latest"
        model = mlflow.sklearn.load_model(model_uri)
        print(f"Successfully loaded production model: {model_uri}")
    except Exception:
        # Fallback estimator for testing without an active MLflow tracking server
        from sklearn.ensemble import GradientBoostingRegressor
        model = GradientBoostingRegressor().fit(
            np.random.rand(100, 10), 
            np.random.randint(500, 5000, 100)
        )
        print("MLflow server offline. Initialized fallback model instance for demonstration.")
    yield
    # Code after 'yield' executes on shutdown (if needed)

# Initialize FastAPI app with lifespan
app = FastAPI(
    title="Traffic Volume Prediction Service",
    description="MLOps Serving API for Corridor Travel Recommendation Engine",
    version="3.0.0",
    lifespan=lifespan
)

# ==============================================================================
# 2. MODERN PYDANTIC V2 SCHEMA (Replaces 'example' with 'examples')
# ==============================================================================
class FeaturePayload(BaseModel):
    hour_sin: float = Field(..., examples=[0.5])
    hour_cos: float = Field(..., examples=[-0.866])
    dow_sin: float = Field(..., examples=[0.781])
    dow_cos: float = Field(..., examples=[0.623])
    is_weekend: int = Field(..., ge=0, le=1, examples=[0])
    is_holiday: int = Field(..., ge=0, le=1, examples=[0])
    is_severe_weather: int = Field(..., ge=0, le=1, examples=[0])
    is_low_visibility: int = Field(..., ge=0, le=1, examples=[0])
    temp_scaled: float = Field(..., examples=[0.45])
    clouds_scaled: float = Field(..., examples=[0.20])

class BatchPredictionInput(BaseModel):
    samples: List[FeaturePayload]

# ==============================================================================
# 3. ENDPOINTS
# ==============================================================================
@app.get("/health")
def health_check():
    return {
        "status": "healthy",
        "model_loaded": model is not None,
        "model_name": MODEL_NAME
    }

@app.post("/predict")
def predict_traffic_volume(payload: BatchPredictionInput):
    if model is None:
        raise HTTPException(status_code=500, detail="Model is not initialized.")
    
    # Use Pydantic V2 .model_dump() instead of deprecated .dict()
    data_dicts = [sample.model_dump() for sample in payload.samples]
    input_df = pd.DataFrame(data_dicts)
    
    predictions = model.predict(input_df)
    
    results = []
    for pred in predictions:
        pred_val = int(max(0, pred))
        risk_level = "HIGH" if pred_val > 4000 else "NORMAL"
        results.append({
            "predicted_traffic_volume": pred_val,
            "risk_status": risk_level
        })
        
    return {"predictions": results}
```
