```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
# 1. Load dataset
df = pd.read_csv("data/processed/cleaned_traffic_features.csv")
```

```python
# 2. Select key features for clustering traffic conditions
cluster_cols = [
    'traffic_volume', 'hour', 'day_of_week',
    'is_weekend', 'is_severe_weather', 'is_low_visibility'
]

X_cluster = df[cluster_cols].copy()

# Standardize features for distance-based K-Means algorithm
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_cluster)
```

```python
# 3. Fit K-Means with k=4
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X_scaled)
```

```python
# 4. Generate Cluster Profile Summary
cluster_profile = df.groupby('cluster').agg(
    Avg_Traffic_Volume=('traffic_volume', 'mean'),
    Avg_Hour=('hour', 'mean'),
    Weekend_Pct=('is_weekend', lambda x: np.round(x.mean() * 100, 1)),
    Severe_Weather_Pct=('is_severe_weather', lambda x: np.round(x.mean() * 100, 1)),
    Low_Visibility_Pct=('is_low_visibility', lambda x: np.round(x.mean() * 100, 1)),
    Record_Count=('traffic_volume', 'count')
).reset_index()
```

```python
print("=== K-MEANS CLUSTER PROFILES ===")
print(cluster_profile.to_string(index=False))
```
